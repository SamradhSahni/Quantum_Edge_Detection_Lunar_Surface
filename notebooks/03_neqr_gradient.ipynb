# 03 — NEQR Encoding + Quantum Gradient (Zhang et al. 2013)

Basis-state encoding with exact 8-bit intensities; the horizontal neighbour image is written into a second register via inverted-control preparation; a CDKM ripple-carry subtractor computes F − G (mod 256) exactly.

```python
import sys, os
sys.path.insert(0, os.path.dirname(os.path.abspath('')) if os.path.basename(os.getcwd())!='Quantum_Image' else '.')
sys.path.insert(0, '..') if os.path.basename(os.getcwd())!='Quantum_Image' else None
sys.path.insert(0, os.getcwd())
%matplotlib inline
import numpy as np
from data.synthetic.generator import BUILDERS
from quantum.neqr_encoder import build_neqr_image_circuit, neqr_layout
from quantum.quantum_gradient import build_neqr_gradient_circuit, neqr_gradient_payload
img, _ = BUILDERS['circle'](8)
img255 = np.clip(np.rint(img * 255), 0, 255)
lay = neqr_layout(3)
print('qubit layout @ 8x8: rows', lay['row'], '| cols', lay['col'], '| F', lay['f'], '| G', lay['g'], '| anc', lay['anc'])
qc = build_neqr_gradient_circuit(img255, 'horizontal', measure=True)
print('circuit: qubits', qc.num_qubits, '| ops', dict(qc.count_ops()))
```

```python

from quantum.quantum_pipeline import run_quantum
res, rec = run_quantum(img, 'neqr', shots=8192)
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 3, figsize=(9, 3))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('input')
ax[1].imshow(res.strength, cmap='magma'); ax[1].set_title('NEQR-gradient strength')
ax[2].imshow(apply_threshold(res.strength, 0.2), cmap='gray'); ax[2].set_title('binary')
from quantum.threshold_selector import apply_threshold
for a in ax: a.set_xticks([]); a.set_yticks([])
plt.tight_layout(); plt.savefig('../results/figures/nb03_neqr.png', dpi=130)
```
